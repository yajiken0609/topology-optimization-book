```python
import numpy as np
x = np.array([1.0,  2.0]) # 設計変数の初期解
s = lam_s = np.ones(1) # スラック変数およびラグランジュ乗数の初期解
rho = 1.0 # バリアパラメータの初期値
RHO_TOL = 1e-7 # 収束判定のしきい値
TAU_TOL = 0.99 # ステップ幅の最大値を決めるパラメータ
KKT_COEF = 0.9 # 内部反復のしきい値を決めるパラメータ
RHO_DECAY = 0.1 # バリアパラメータの減少率
STEP_DECAY = 0.5 # ステップ幅の減少率
ITR_MAX = 100 # 外部反復の最大反復数
ind_z = np.cumsum([len(x), len(s), len(lam_s)])[:-1] # 解のインデックス
def compute_kkt(g, gradf, gradg, lam_s, s, rho):
    kkt = np.concatenate([
        gradf+lam_s*gradg, g+s, lam_s*s-rho])
    return kkt # KKT条件の等式をまとめたベクトル
def eval_func(x):
    f = (x[0]-1)**4+(x[0]-2*x[1])**2
    g = x[0]**2-x[1]
    return f, g # 評価関数
def grad(x):
    gradf = np.array([4*(x[0]-1)**3+2*(x[0]-2*x[1]), -4*(x[0]-2*x[1])])
    gradg = np.array([2*x[0], -1])
    return gradf, gradg # 評価関数の勾配
def Hessian(x):
    Hf = np.array([[12*(x[0]-1)**2+2, -4], [-4, 8]])
    Hg = np.array([[2, 0], [0, 0]])
    return Hf, Hg # 評価関数のヘッセ行列
trajectory = [x.copy()] # 設計変数の軌跡
for itr in range(ITR_MAX):
    f, g = eval_func(x)
    gradf, gradg = grad(x)
    tmp_kkt = compute_kkt(g, gradf, gradg, lam_s, s, rho)
    norm_l2_kkt = np.linalg.norm(tmp_kkt, 2) # ユークリッドノルム
    norm_inf_kkt = np.linalg.norm(tmp_kkt, np.inf) # 最大値ノルム
    print(f"{itr} iters, x = ({x[0]:.3f}, {x[1]:.3f}), rho = {rho:.1e}, norm_inf_kkt = {norm_inf_kkt:.1e}")
    if rho < RHO_TOL: break # 外部反復の収束判定
    while norm_inf_kkt > KKT_COEF*rho: # 内部反復
        _, g = eval_func(x)
        gradf, gradg = grad(x)
        Hf, Hg1 = Hessian(x)
        J = np.zeros((4, 4)) # ニュートン法のためのヤコビ行列
        J[np.ix_(np.array([0, 1]),np.array([0, 1]))] += Hf+lam_s*Hg1
        J[np.ix_(np.array([3, 3]),np.array([0, 1]))] += gradg
        J[np.ix_(np.array([0, 1]),np.array([3, 3]))] += gradg[:, None]
        J[3, 2] += 1
        J[2, 3] += s[0]
        J[2, 2] += lam_s[0]
        r_rho = np.zeros(4) # ニュートン法のための右辺ベクトル
        r_rho[[0, 1]] -= gradf+lam_s[0]*gradg
        r_rho[2] -= lam_s[0]*s[0]-rho
        r_rho[3] -= g+s[0]
        dz = np.linalg.solve(J, r_rho) # 連立方程式の求解
        dx, ds, dlam_s = np.split(dz, ind_z) 
        inv_t = np.maximum(-ds/(TAU_TOL*s), -dlam_s/(TAU_TOL*lam_s))
        t = 1/np.maximum(inv_t, 1) # 最大許容ステップ幅
        x_old, s_old, lam_s_old = x.copy(), s.copy(), lam_s.copy()
        tmp_norm_l2_kkt = 2*norm_l2_kkt            
        while tmp_norm_l2_kkt > norm_l2_kkt: # 残差ノルムによる直線探索
            x, s, lam_s = x_old+t*dx, s_old+t*ds, lam_s_old+t*dlam_s
            _, g = eval_func(x)
            gradf, gradg = grad(x)
            tmp_kkt = compute_kkt(g, gradf, gradg, lam_s, s, rho)
            tmp_norm_l2_kkt = np.linalg.norm(tmp_kkt, 2)                
            t = STEP_DECAY*t
        norm_l2_kkt = tmp_norm_l2_kkt
        norm_inf_kkt = np.linalg.norm(tmp_kkt, np.inf)
    rho = RHO_DECAY*rho # バリアパラメータの更新
    trajectory.append(x.copy()) # 設計変数の軌跡を記録
```

```python
import matplotlib.pyplot as plt
x_p = np.linspace(0, 3, 400)
y_p = np.linspace(0, 3, 400)
X, Y = np.meshgrid(x_p, y_p)
Z, G1 = eval_func(np.array([X, Y]))
feasible_region = (G1 < 0) 
infeasible_region = (G1 > 0) 
contour = plt.contourf(X, Y, Z, levels=40, cmap="viridis", alpha=0.7)
plt.colorbar(label='$f$')
plt.contourf(X, Y, infeasible_region, levels=[0.5, 1], colors='white', alpha=1.0)
constraint1 = plt.contour(X, Y, G1, levels=[0], linestyles='--', colors='black', linewidths=2)
plt.contour(X, Y, Z, levels=40, linestyles='-', linewidths=1, colors='gray', alpha=1)
traj_x = np.squeeze(trajectory)
plt.plot(traj_x[:, 0], traj_x[:, 1], color='k', marker='.', markersize=10, label="Trajectory")
plt.plot(traj_x[0, 0], traj_x[0, 1], color='k', marker='.', linestyle="", markersize=20, markeredgecolor='w', markeredgewidth=2, label="Initial")
plt.plot(traj_x[-1, 0], traj_x[-1, 1],color='k', marker='*', linestyle="", markersize=20, markeredgecolor='w', markeredgewidth=2, label="Optimized")
plt.xlabel("$x_1$")
plt.ylabel("$x_2$")
plt.gca().set_aspect('equal')
plt.legend(framealpha=1)
```
